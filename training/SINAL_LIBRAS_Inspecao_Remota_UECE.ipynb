# SINAL LIBRAS — inspeção remota do pacote contínuo UECE

Este notebook **não baixa o ZIP de 35,6 GB inteiro**. Ele tenta ler o diretório central e cabeçalhos NPY por HTTP Range. O servidor precisa aceitar esse tipo de requisição. Se não aceitar, o processo para sem iniciar um download completo.

Use **Ambiente de execução → Executar tudo**. O processamento é feito no ambiente gratuito do Colab, sujeito aos limites e à disponibilidade do Google. Não envia vídeos pessoais.

**Segurança:** o notebook não usa `pickle.load` nem `numpy.load(..., allow_pickle=True)`. O dataset tem licença CC BY-NC 4.0; não pode ser usado em produto comercial sem outra autorização.

```python
!pip -q install remotezip requests
from pathlib import Path
import json, subprocess, sys, urllib.request

WORK = Path('/content/sinal_libras_remote_audit')
WORK.mkdir(exist_ok=True)
script_url = 'https://raw.githubusercontent.com/danyellallk/SINAL-LIBRAS/main/training/inspect_uece_remote_zip.py'
script = WORK / 'inspect_uece_remote_zip.py'
urllib.request.urlretrieve(script_url, script)
print('Ferramenta preparada:', script)
```

```python
# Inspeção remota: somente metadados e até 80 cabeçalhos NPY.
# Se HTTP Range não estiver disponível, o script encerra sem baixar o ZIP completo.
result = subprocess.run(
    [sys.executable, str(script), '--max-samples', '80'],
    text=True, capture_output=True, timeout=1800
)
print(result.stdout)
if result.stderr:
    print('MENSAGENS:', result.stderr)
print('Código de saída:', result.returncode)
report = WORK / 'relatorio_inspecao_remota.txt'
report.write_text('SAÍDA:\n' + result.stdout + '\nMENSAGENS:\n' + result.stderr + '\nCÓDIGO: ' + str(result.returncode), encoding='utf-8')
print('Relatório salvo em:', report)
```

```python
# Baixa o relatório pequeno para compartilhar aqui.
from google.colab import files
files.download(str(report))
```
